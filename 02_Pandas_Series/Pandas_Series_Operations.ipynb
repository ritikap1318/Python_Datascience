### What is Pandas

Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool,

### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

```python
import numpy as np
import pandas as pd
```

Series from lists

```python
#string
country = ['India','Pakistan','USA','Nepal','Srilanka']
pd.Series(country)
```

```python
#integers
runs = [13,24,56,78,100]

runs_ser = pd.Series(runs)
```

```python
#we can provide index the names as well
marks=[67,68,69,100,90]
subject=['Maths','SST','Science','English','Hindi']
marks_ser = pd.Series(marks,index=subject)
marks_ser
```

```python
marks_ser=pd.Series(marks,index=subject,name='Ritika ke marks')
marks_ser
```

### Series from dict


```python
marks = {
    'maths':67,
    'sst':68,
    'science':69,
    'english':100,
    'hindi':90
}
marks_series=pd.Series(marks,name='ritika marks')
marks_series
```
